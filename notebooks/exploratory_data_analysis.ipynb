### 3. Conversion Rate by Primary Device
Analyzing purchase conversion across Smartphones, Laptops/Desktops, and Tablets.

```python
device_conv = df.groupby('Primary_Device')[TARGET_COLUMN].apply(lambda s: (s == 'Yes').mean() * 100).reset_index(name='Conversion_Rate_Pct')
print(device_conv)

plt.figure(figsize=(7, 4), dpi=150)
sns.barplot(data=device_conv, x='Primary_Device', y='Conversion_Rate_Pct', palette='Blues_d')
plt.title('Conversion Rate by Device Type (%)')
plt.ylabel('Conversion Rate (%)')
plt.show()
```

### 4. Cart Page Value Score vs. Completed Purchase
Evaluating the commercial intent score distinction between buyers and non-buyers.

```python
plt.figure(figsize=(7, 4), dpi=150)
sns.boxplot(data=df, x=TARGET_COLUMN, y='Cart_Page_Value_Score', palette=['#94A3B8', '#16A34A'])
plt.title('Cart Page Value Score Distribution by Purchase Outcome')
plt.xlabel('Completed Purchase')
plt.ylabel('Cart Page Value Score ($)')
plt.show()
```
